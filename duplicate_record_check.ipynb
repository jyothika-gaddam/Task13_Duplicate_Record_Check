```python
import pandas as pd

# Load the dataset
df = pd.read_csv("../data/superstore_raw.csv", sep="\t")

# Display basic information
print("Original records:", len(df))
print("Original columns:", len(df.columns))

# Check duplicate records
duplicate_count = df.duplicated().sum()

print("Duplicate records:", duplicate_count)

# Remove duplicate records
df_cleaned = df.drop_duplicates()

print("Records after removing duplicates:", len(df_cleaned))

# Save cleaned dataset
df_cleaned.to_csv("../data/superstore_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```

```python
# Display duplicate records

duplicates = df[df.duplicated(keep=False)]

print("Total duplicate records:", len(duplicates))

duplicates.head(10)
```

```python
# Calculate duplicate percentage

duplicate_percentage = (df.duplicated().sum() / len(df)) * 100

print(f"Duplicate percentage: {duplicate_percentage:.2f}%")
```

```python
# Final verification

print("Original records:", len(df))
print("Duplicate records:", df.duplicated().sum())
print("Cleaned records:", len(df_cleaned))
print("Records removed:", len(df) - len(df_cleaned))
```
